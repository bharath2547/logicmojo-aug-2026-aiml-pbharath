### Problem-1

In a network, data packets are sent through a shared channel. Suppose there are 10 devices attempting to send data. In any given time slot, each device has a 20% chance of attempting to send a packet, independently of other devices. A collision occurs if two or more devices attempt to send a packet in the same time slot. What is the probability that in a given time slot, there will be no collisions?

### Solution

Let $N$ be the number of devices that transmit. There is no collision precisely when $N=0$ or $N=1$. By independence, $N\sim\operatorname{Binomial}(10,0.2)$, so

$$
P(\text{no collision})=P(N=0)+P(N=1)
=(0.8)^{10}+\binom{10}{1}(0.2)(0.8)^9
=0.3758096384.
$$

**Answer: approximately $0.37581$ (37.58%).**

### Problem-2

A system uses 8-character passwords. Each character can be any lowercase letter (a-z), uppercase letter (A-Z), or digit (0-9).
What is the probability that a randomly generated 8-character password contains at least one digit, given that character repetition is allowed?

### Solution

There are 62 possible characters, of which 52 are letters and 10 are digits. The probability that all 8 characters are letters is $(52/62)^8$, because repetition is allowed and character choices are independent. Therefore, by the complement rule,

$$
P(\text{at least one digit})=1-\left(\frac{52}{62}\right)^8
\approx 0.99904357.
$$

**Answer: approximately 0.99904 (99.904%).**

### Problem-3

A software project has two modules, A and B. The probability that module A has a bug is 0.15, and the probability that module B has a bug is 0.10. The probability that both modules have bugs is 0.03.

(a) What is the probability that module A has a bug, given that module B has a bug?

(b) What is the probability that module B has a bug, given that module A does not have a bug?

### Solution

Use the conditional-probability formula.

**(a)**
$$
P(A\mid B)=\frac{P(A\cap B)}{P(B)}=\frac{0.03}{0.10}=0.30.
$$

**(b)** First, $P(B\cap A^c)=P(B)-P(A\cap B)=0.10-0.03=0.07$, and $P(A^c)=1-0.15=0.85$. Thus
$$
P(B\mid A^c)=\frac{0.07}{0.85}\approx 0.082353.
$$

**Answers: (a) 0.30; (b) approximately 0.08235.**

### Problem-4

A robot uses a sensor to detect an obstacle. The sensor has a 95% chance of correctly detecting an obstacle when one is present (true positive rate). However, it also has a 2% chance of indicating an obstacle when there isn't one (false positive rate).
In the robot's environment, obstacles are present 10% of the time.

What is the probability that an obstacle is truly present, given that the sensor indicates an obstacle?

### Solution

Let $O$ denote an obstacle being present and $+$ denote a positive sensor indication. Bayes' theorem gives

$$
P(O\mid +)=\frac{P(+\mid O)P(O)}{P(+\mid O)P(O)+P(+\mid O^c)P(O^c)}
=\frac{0.95(0.10)}{0.95(0.10)+0.02(0.90)}
=\frac{0.095}{0.113}\approx 0.840708.
$$

**Answer: approximately 84.07%.**

### Problem-5

A spam filter uses a keyword "money" to identify spam emails.
- 10% of all emails are spam.
- If an email is spam, the word "money" appears in it 80% of the time.
- If an email is not spam, the word "money" appears in it 5% of the time.

If an email contains the word "money", what is the probability that it is spam?

### Solution

Let $S$ mean spam and $M$ mean that the email contains “money.” By Bayes' theorem,

$$
P(S\mid M)=\frac{P(M\mid S)P(S)}{P(M\mid S)P(S)+P(M\mid S^c)P(S^c)}
=\frac{0.80(0.10)}{0.80(0.10)+0.05(0.90)}
=\frac{0.08}{0.125}=0.64.
$$

**Answer: 0.64 (64%).**

### Problem-9

The noise in a sensor's reading can be modeled by a Gaussian (Normal) distribution with a mean of 0 (bias) and a standard deviation of 0.5 units.

For latency $X\sim N(150,20^2)$ ms, standardize with $Z=(X-150)/20$.

**(a)** $180$ ms corresponds to $z=(180-150)/20=1.5$, so

**(b)** The 90th percentile has $z_{0.90}\approx 1.2815516$:
$$x_{0.90}=150+20(1.2815516)\approx 175.631\text{ ms}.$$

**(c)** Critical latency is above $150+2.5(20)=200$ ms, hence
$$P(Z>2.5)=1-\Phi(2.5)\approx 0.00620967.$$

**Answers: (a) 6.68%; (b) approximately 175.63 ms; (c) approximately 0.621%.**

### Problem-12
The probability density function (pdf) of the triangular distribution is characterized by three points as shown in the figure below.

<img src="https://upload.wikimedia.org/wikipedia/commons/4/45/Triangular_distribution_PMF.png" alt="triangular distribution" width="500"/>

(a) min value of the distribution,
(b) max value of the distribution, and
(c) mode of the distribution

Note: $a \leq c \leq b$

Create a class `TriangularDistribution` with attributes `min`, `max` and `mode`. You need to implement the following methods:

- `pdf(x)`: returns the value of the pdf at point $x$.
- `plot_pdf()`: plots the pdf in the range $[min , max]$
- `cdf(x)`: returns the value of the cdf at point $x$ ; which denotes $P(X \leq x)$
- `plot_cdf()`: plots the cdf in the range $[min , max]$
- `get_probability(x1, x2)`: returns the probability $P(x_1 \leq X \leq x_2)$, where $x_1 \lt x_2$.

Gracefully handle the possible errors and also the edge conditions when either $ a = c $ or $b = c$ in the distribution.

Test your implementation.

````python
import math
import numbers
import numpy as np
import matplotlib.pyplot as plt

class TriangularDistribution:
    """Triangular distribution with minimum, maximum, and mode parameters."""

    def __init__(self, min, max, mode):
        values = (min, max, mode)
        if any(not isinstance(v, numbers.Real) or not math.isfinite(float(v)) for v in values):
            raise TypeError("min, max, and mode must be finite real numbers")
        self.min = float(min)
        self.max = float(max)
        self.mode = float(mode)
        if self.min >= self.max:
            raise ValueError("min must be strictly less than max")
        if not self.min <= self.mode <= self.max:
            raise ValueError("mode must lie between min and max, inclusive")

    def pdf(self, x):
        """Probability density at x (zero outside the support)."""
        x = float(x)
        if not math.isfinite(x):
            raise ValueError("x must be finite")
        a, b, c = self.min, self.max, self.mode
        if x < a or x > b:
            return 0.0
        # When the mode is an endpoint, only one of the two usual pieces remains.
        if c == a:
            return 2.0 * (b - x) / ((b - a) ** 2)
        if c == b:
            return 2.0 * (x - a) / ((b - a) ** 2)
        if x <= c:
            return 2.0 * (x - a) / ((b - a) * (c - a))
        return 2.0 * (b - x) / ((b - a) * (b - c))

    def cdf(self, x):
        """Cumulative probability P(X <= x)."""
        x = float(x)
        if not math.isfinite(x):
            raise ValueError("x must be finite")
        a, b, c = self.min, self.max, self.mode
        if x <= a:
            return 0.0
        if x >= b:
            return 1.0
        if c == a:
            return 1.0 - ((b - x) ** 2) / ((b - a) ** 2)
        if c == b:
            return ((x - a) ** 2) / ((b - a) ** 2)
        if x <= c:
            return ((x - a) ** 2) / ((b - a) * (c - a))
        return 1.0 - ((b - x) ** 2) / ((b - a) * (b - c))

    def plot_pdf(self):
        """Plot the density on [min, max]; return (figure, axes)."""
        xs = np.linspace(self.min, self.max, 500)
        ys = [self.pdf(x) for x in xs]
        fig, ax = plt.subplots()
        ax.plot(xs, ys, label="PDF")
        ax.set(xlabel="x", ylabel="Density", title="Triangular distribution PDF")
        ax.grid(True, alpha=0.3)
        ax.legend()
        return fig, ax

    def plot_cdf(self):
        """Plot the cumulative distribution on [min, max]; return (figure, axes)."""
        xs = np.linspace(self.min, self.max, 500)
        ys = [self.cdf(x) for x in xs]
        fig, ax = plt.subplots()
        ax.plot(xs, ys, label="CDF")
        ax.set(xlabel="x", ylabel="P(X <= x)", title="Triangular distribution CDF")
        ax.set_ylim(0, 1)
        ax.grid(True, alpha=0.3)
        ax.legend()
        return fig, ax

    def get_probability(self, x1, x2):
        """Return P(x1 <= X <= x2), with x1 strictly less than x2."""
        x1, x2 = float(x1), float(x2)
        if not math.isfinite(x1) or not math.isfinite(x2):
            raise ValueError("interval endpoints must be finite")
        if x1 >= x2:
            raise ValueError("x1 must be strictly less than x2")
        # A continuous distribution assigns zero probability to individual endpoints.
        return max(0.0, self.cdf(x2) - self.cdf(x1))

# Tests: an interior mode and both endpoint-mode edge cases.
tri = TriangularDistribution(0, 10, 4)
assert math.isclose(tri.pdf(4), 2 / 10)
assert math.isclose(tri.cdf(4), 0.4)
assert math.isclose(tri.get_probability(0, 10), 1.0)
assert tri.pdf(-1) == 0.0 and tri.cdf(-1) == 0.0 and tri.cdf(11) == 1.0
left_mode = TriangularDistribution(0, 10, 0)
right_mode = TriangularDistribution(0, 10, 10)
assert math.isclose(left_mode.pdf(0), 0.2) and math.isclose(left_mode.cdf(5), 0.75)
assert math.isclose(right_mode.pdf(10), 0.2) and math.isclose(right_mode.cdf(5), 0.25)
for bad_args in [(1, 1, 1), (0, 1, 2)]:
    try:
        TriangularDistribution(*bad_args)
    except ValueError:
        pass
    else:
        raise AssertionError("Invalid distribution parameters should raise ValueError")
print("TriangularDistribution tests passed.")
````

### Problem-13

The `rand5()` function which returns a uniform random integer within the range [1, 5], is defined as following:

```python
def rand5():
    import numpy as np
    return np.random.randint(1,6)
```

Write a function `rand7()` that will generate a uniform random integer within the range [1, 7] by only using `rand5()`. 

Note: You can call `rand5()` as many time as you want inside the function `rand7()`, but you can not use `np.random.randint(1, 8)` or any other built-in methods in python. 

````python
import numpy as np

def rand5():
    """Uniform random integer from 1 through 5, as specified in the question."""
    return np.random.randint(1, 6)

def rand7():
    """Uniform random integer from 1 through 7, using only rand5()."""
    while True:
        # Two independent rand5 calls make each integer 1..25 equally likely.
        value = (rand5() - 1) * 5 + rand5()  # uniform on {1, ..., 25}
        if value <= 21:                       # reject the four excess outcomes
            return (value - 1) % 7 + 1

# Basic validity test; the construction is exactly uniform because each
# accepted output has the same number (three) of preimages among 1..21.
samples = [rand7() for _ in range(1000)]
assert all(1 <= value <= 7 for value in samples)
print("rand7 validity test passed; observed values:", sorted(set(samples)))
````

### Problem-14

A grayscale image can be represented as a matrix where the dimensions are (height, width). 
Let's consider a simplified scenario with two grayscale images, $A$ and $B$, both represented as 10×10 matrices where each element represents pixel intensity. Pixel intensity is an integer in the range [0-255].

You want to create a blended image $C$ by combining $A$ and $B$ such that $C=w_1 A+w_2 B$. This represents a weighted average blend.
Note that each element in the resultant matrix $C$ should also fall within the range [0, 255]. 

write the code of the function `blend_images(image_A, image_B, weight_A, weight_B)` that takes two matrices and two scalar weights and produces another matrix as an output by performing the above operation.

````python
import numpy as np

def blend_images(image_A, image_B, weight_A, weight_B):
    """Blend two grayscale matrices and return uint8 pixels in [0, 255].

    Weights must be finite and nonnegative. Values outside the pixel range
    after blending are clipped; fractional intensities are rounded to nearest.
    """
    A = np.asarray(image_A)
    B = np.asarray(image_B)
    if A.ndim != 2 or B.ndim != 2:
        raise ValueError("Both images must be 2D grayscale matrices")
    if A.shape != B.shape:
        raise ValueError("Both images must have the same dimensions")
    if not np.issubdtype(A.dtype, np.number) or not np.issubdtype(B.dtype, np.number):
        raise TypeError("Image pixels must be numeric")
    if not np.all(np.isfinite(A)) or not np.all(np.isfinite(B)):
        raise ValueError("Image pixels must be finite")
    if np.any((A < 0) | (A > 255)) or np.any((B < 0) | (B > 255)):
        raise ValueError("Input pixel intensities must be in [0, 255]")
    wA, wB = float(weight_A), float(weight_B)
    if not np.isfinite(wA) or not np.isfinite(wB) or wA < 0 or wB < 0:
        raise ValueError("Weights must be finite and nonnegative")

    blended = wA * A.astype(float) + wB * B.astype(float)
    return np.clip(np.rint(blended), 0, 255).astype(np.uint8)

# Test normal weighted blending and the required output range/type.
A = np.array([[0, 100], [200, 255]], dtype=np.uint8)
B = np.array([[255, 200], [100, 0]], dtype=np.uint8)
C = blend_images(A, B, 0.25, 0.75)
assert np.array_equal(C, np.array([[191, 175], [125, 64]], dtype=np.uint8))
assert C.dtype == np.uint8 and np.all((C >= 0) & (C <= 255))
print("blend_images tests passed; result:\n", C)
````

### Problem-15

Calculate the eigen values and eigen vectors of the following matrices.

$A = \begin{bmatrix}
5 & 2 \\
2 & 2
\end{bmatrix}$      

$B = \begin{bmatrix}
1 & 1 & 0\\
0 & 2 & 0\\
-1 & 1 & 3
\end{bmatrix}$

### Solution

For matrix $A$, the characteristic polynomial is $\lambda^2-7\lambda+6=(\lambda-6)(\lambda-1)$. Thus the eigenvalues are $6$ and $1$. Corresponding eigenvectors are $(2,1)^T$ and $(-1,2)^T$, respectively (any nonzero scalar multiples are also valid).

For matrix $B$, the characteristic polynomial is $(1-\lambda)(2-\lambda)(3-\lambda)$, so the eigenvalues are $1,2,3$. Corresponding eigenvectors are $(2,0,1)^T$, $(1,1,0)^T$, and $(0,0,1)^T$, respectively.

### Numerical verification

The following cell uses NumPy to verify the eigenpairs. Eigenvectors returned by a numerical library may have different signs or normalization; those are equivalent eigenvectors.

````python
import numpy as np
A = np.array([[5, 2], [2, 2]], dtype=float)
B = np.array([[1, 1, 0], [0, 2, 0], [-1, 1, 3]], dtype=float)
for name, matrix in (("A", A), ("B", B)):
    eigenvalues, eigenvectors = np.linalg.eig(matrix)
    print(f"Matrix {name} eigenvalues:\n", eigenvalues)
    print(f"Matrix {name} eigenvectors (columns correspond to eigenvalues):\n", eigenvectors)
    assert np.allclose(matrix @ eigenvectors, eigenvectors @ np.diag(eigenvalues))
print("Eigenpair checks passed.")
````
