## Problem: 4

Create a 10 x 10 arrays of zeros and then "frame" it with a border of ones.

## Problem: 6

Create an array $x$ with 100 points in the interval $[-1,1]$ (the points must be uniformly spaced). Then create another array $y$. 

Where $y = x^2 sin(x^2) + x$. 

Try to plot $y$ vs $x$ using matplotlib.

```python
# Write your code here
import matplotlib.pyplot as plt

x = np.linspace(start=-1, stop=1, num=100)
#print(x)

x_squared = x**2

y = x_squared * np.sin(x_squared) + x
#print(y)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label="y = x² * sin(x²) + x", color="blue", linewidth=2)

plt.title("Plot of y versus x", fontsize=14, pad=12)
plt.xlabel("x", fontsize=12)
plt.ylabel("y", fontsize=12)

plt.axhline(0, color="gray", linewidth=0.8, linestyle="--")
plt.axvline(0, color="gray", linewidth=0.8, linestyle="--")

plt.grid(True, alpha=0.3)
plt.legend(loc="upper left")

plt.show()
```

## Problem: 7 
Create a function which creates an $n×n$ array with $(i , j)$ -entry equal to $i+ j$

```python
# Write your code here
def create_sum_matrix(n):
    h_arr = np.arange(n).reshape(1, n)
    v_arr = h_arr.reshape(n, 1)

    return h_arr + v_arr

print(create_sum_matrix(4))
```

# Problem: 8
Evaluate $cos$ and $sin$ on the interval $[0,1]$  with 1000 points and then stack the results into a tall array with rows being the $(cos(x),sin(x))$ entries.

```python
# Write your code here
x = np.linspace(start=0, stop=1, num=1000)
sin_x = np.sin(x)
cos_x = np.cos(x)
print(f"sin(x) shape: {sin_x.shape}"  )
print(f"cos(x) shape: {cos_x.shape}")

tall_array = np.stack((cos_x, sin_x), axis=1)
print(tall_array.shape)
```

# Problem: 9

Create a random $3×5$ array using the `np.random.rand(3 ,5)` function and compute: the sum of all the entries, the sum of the rows, and the sum of the columns.

```python
# Write your code here
nums = np.random.rand(3, 5)

print(nums)
print(f"Sum of all entries: {np.sum(nums)}")
print(f"Sum of rows: {np.sum(nums, axis=1)}")
print(f"Sum of cols: {np.sum(nums, axis=0)}")
```
